## Definition :
Data types are a classification of data which tell the compiler or interpreter how the programmer intends to use the data.
They determine the type of operations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data.
## Importance of DataTypes in programming
- Data Types ensure that data is stored in an efficient way.
- They help in performing correct operations on data.
- Proper use of data types can prevent errors and bugs in the program.


## what is Type Conversion?
- Type conversion is needed to convert data from one type to another
why do we need Type conversion 

## Why do we need Type Conversion?
- We can perform correct operations, avoid errors, and handle user input or real-world data efficiently.

## Types of Type Conversion?

   1. Implicit Type conversion :Automatical

   2. Explicit Type conversion:manually we convert them.

- Mostly we use datatypes for type conversion are:
int(), float(), str()

```python
 ##3.Typecasting: Type Conversion means changing one data type into another data type.

result="Hello"+str(5)
print(result)
'''
in the above example string cant be converted into the integer 
bcz by default it consider other also as a string so for that you need to \
do typeCasting i.e by add str(5) before the integer add str'''
```

```python
          ## 1. Implict type conversion:Automatic
num = input("Enter number: ")   # "1001"
print(num + num)   # "10011001" (wrong)

num = int(num)
print(num + num)   # 2002 (correct)
```

```python
## without conversion Results may be wrong
price = "100.5"

# Wrong
print(price * 2)   # "100.5100.5" 

# Correct
price = float(price)
print(price * 2)   # 201.0 
```

```python
## string to float 
price = input("Enter price: ")   # string

price = float(price)             # convert to float

total = price * 2
print("Total:", total)
```

```python
            ##2.Explicit type casting: Done manually by programmer
x = "100"

# converting string to int
y = int(x)

print(y)        # 100
print(type(y))  # int
```
