## Sets 
- sets are a built in data type in python used to store collections of unique items.They are unordered,meaning that the elements dont follow a specific order,and they dont allow duplicate elements.Sets are useful for membership tests,eliminating duplicate entries and perfirming mathemetical set operations like union ,intersection,difference and su=ymmetric difference.

```python
## how to create a Sets
my_set= {1,2,3,4,5}
print(type(my_set))
print(my_set)
```

```python
my_empty_set= set([1,2,3,4,5])
print(type(my_empty_set))
print(my_empty_set)
```

```python
##Basic set Operation
## Adding and Removing Elements 
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## Removing Elements 
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
print(my_set)
```

```python
my_set.discard(11)
print(my_set)
```

```python
##removing a random element using pop
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
##clear all the elements
my_set.clear()
print(my_set)
```

```python
## set membership test 
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
##Mathematical Operation 
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
## Union 
union_set=set1.union(set2)
print(union_set)
##Intersection 
intersection_set=set1.intersection(set2)
print(intersection_set)
set1.intersection_update(set2) ## findout the common elements 
print(set1)
```

```python
##Difference
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
print(set1.difference(set2))
```

```python
##Symmetric Difference 
set1.symmetric_difference(set2)
```

```python
## sets Methods 
set1={1,2,3}
set2={3,4,5}
## is subset
print(set1.issubset(set2))
## set1 is superset of set2

print(set1.issuperset(set2))
```
