## 1. Syntax in Python

**Syntax** refers to the **rules that define the correct structure of a Python program**.  
It tells us **how to write statements and code correctly so that Python can understand it**.

If the syntax is incorrect, Python will generate a **SyntaxError**.

## Syntax Rules in Python

Python is case-sensitive

Indentation is mandatory

Statements usually end at a new line

Parentheses, colons, and quotes must be used correctly

````python
a = 10
b = "5"

print(a + b)
#The syntax is correct, but the meaning of the operation is wrong because we cannot add an integer and a string.
````
