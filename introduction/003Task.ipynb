```python
## We doing this task to understand :
## In Python, the input() function always returns data as a string.
## float(): To convert the string input into a numeric type.
## float Instead of int :We use float so the calculator can accept decimal numbers too.
## Python is dynamically typed, which means:You do not declare the variable type explicitly.Python automatically changes the data type during runtime.
## Flow chart:
'''
User Input
   ↓
input() → returns string
   ↓
float() → converts to number
   ↓
Arithmetic operations
   ↓
Output
'''
num1 = float(input("Enter first Number:"))
num2 = float(input("Enter Second Number:"))
sum = num1 +  num2
difference = num1 - num2
product= num1  * num2
quotient = num1 / num2

print("Sum:",sum)
print("Difference:",difference)
print("Product:",product)
print("Quotient:",quotient)

```
